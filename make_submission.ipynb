```python
import pandas as pd

men_com = pd.read_csv("submission_files/men_com_submission_file.csv")
men_base = pd.read_csv("submission_files/men_base_submission_file.csv")
men_r = pd.read_csv("submission_files/men_r_submission_file.csv")
men_w = pd.read_csv("submission_files/men_w_submission_file.csv")
men_main = pd.read_csv("submission_files/men_main_submission_file.csv")

women_com = pd.read_csv("submission_files/women_com_submission_file.csv")
women_base = pd.read_csv("submission_files/women_base_submission_file.csv")
women_r = pd.read_csv("submission_files/women_r_submission_file.csv")
women_w = pd.read_csv("submission_files/women_w_submission_file.csv")
women_main = pd.read_csv("submission_files/women_main_submission_file.csv")

com_submission_file = pd.concat([men_com, women_com], ignore_index=True)
base_submission_file = pd.concat([men_base, women_base], ignore_index=True)
r_submission_file = pd.concat([men_r, women_r], ignore_index=True)
w_submission_file = pd.concat([men_w, women_w], ignore_index=True)
main_submission = pd.concat([men_main, women_main], ignore_index=True)

com_submission_file.to_csv("submission_files/com_submission_file.csv", index=False, lineterminator='\n')
base_submission_file.to_csv("submission_files/base_submission_file.csv", index=False, lineterminator='\n')
r_submission_file.to_csv("submission_files/r_submission_file.csv", index=False, lineterminator='\n')
w_submission_file.to_csv("submission_files/w_submission_file.csv", index=False, lineterminator='\n')
main_submission.to_csv("submission_files/main_submission_file.csv", index=False, lineterminator='\n')
```
